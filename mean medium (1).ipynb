# data set missing values 

```python
# Original data with missing values (None)
data = [5, 8, None, 7, 10, None, 6]

# Calculate mean of non-missing values
total = 0
count = 0
for x in data:
    if x is not None:
        total += x
        count += 1
mean_val = total / count

# Impute missing values with the calculated mean
imputed = []
for x in data:
    if x is None:
        imputed.append(mean_val)
    else:
        imputed.append(x)

print("Original Data:", data)
print("Imputed Data:", imputed)
print("Mean Value:", mean_val)
```

```python
#Dataset with missing values
data = [5, 8, None, 7, 10, None, 6]
total = 0
count = 0
for x in data:
    if x is not None:
        total += x
        count += 1
mean_val = total / count   #mean of datas which are non-missing values
imputed_data = data[:]   #copy of original list
for i in range(len(imputed_data)):
    if imputed_data[i] is None:
        imputed_data[i] = mean_val   # only update where None
total = 0
count = 0
for x in imputed_data:
    total += x
    count += 1
final_mean = total / count

# sorting without inbuilt statistics function (but using sorted() for ordering)

sorted_data = sorted(imputed_data)
n = len(sorted_data)
if n % 2 == 0:   # even
    mid1 = sorted_data[n//2 - 1]
    mid2 = sorted_data[n//2]
    final_median = (mid1 + mid2) / 2
else:            # odd
    final_median = sorted_data[n//2]
print("Original Data :", data)
print("Imputed Data  :", imputed_data)
print("Mean   :", final_mean)
print("Median :", final_median)
```

```python
d = [4, 5, 7, 3, 7, None, 4, None]
clear_d = [x for x in d if x is not None]
total = 0
count = 0
for x in clear_d:
    total += x
    count += 1
mean1 = total / count if count > 0 else 0
# For median, sort the filtered list
sort_d = sorted(clear_d)
n = len(sort_d)
if n % 2 == 0:
    median = (sort_d[n // 2] + sort_d[n // 2 - 1]) / 2
else:
    median = sort_d[n // 2]
print("Mean:", mean1)
print("median",median)
print("orignal data",d)
```
